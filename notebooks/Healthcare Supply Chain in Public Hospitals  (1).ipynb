```python
import pandas as pd
import numpy as np
import random

calendar = pd.read_csv("Calendar.csv")
hospitals = pd.read_csv("Hospitals.csv")
products = pd.read_csv("Products.csv")
suppliers = pd.read_csv("Suppliers.csv")
```

```python
# ==========================================
# Hospital Stock Factor
# ==========================================

hospital_stock_factor = {

    "Teaching": 2.0,

    "Specialized": 1.6,

    "General": 1.2,

    "Central": 1.0,

    "Fever": 1.4

}
```

```python
# ==================================================
# Generate Inventory
# ==================================================

for _, hospital in hospitals.iterrows():
    for _, product in products.iterrows():

        factor = hospital_stock_factor[hospital["HospitalType"]]

        # 1. تعريف min_stock أولاً
        min_stock = int(product["ReorderLevel"] * factor)

        # 2. حساب base_stock
        if product["DemandLevel"] == "High":
            base_stock = np.random.randint(1200, 2500)
        elif product["DemandLevel"] == "Medium":
            base_stock = np.random.randint(600, 1200)
        else:
            base_stock = np.random.randint(150, 600)

        fill_rate = np.random.uniform(0.8, 1.1)

        # 3. حساب current_stock بعد تعريف min_stock
        current_stock = int(base_stock * factor * fill_rate)
        current_stock = max(current_stock, min_stock + np.random.randint(50, 300))

        max_stock = int(current_stock + np.random.randint(500, 1500))

        # Dates
        manufacturing = pd.Timestamp("2024-01-01") + pd.to_timedelta(
            np.random.randint(0, 365), unit="D"
        )
        expiry = manufacturing + pd.DateOffset(
            months=int(product["ShelfLifeMonths"])
        )

        # Storage
        if product["Category"] == "Vaccines" or "Insulin" in product["ProductName"]:
            storage = "Cold Chain"
            zone = "Cold Room"
        else:
            storage = "Room Temperature"
            zone = random.choice(["A", "B", "C"])

        inventory.append({
            "InventoryID": f"INV{inventory_id:05}",
            "HospitalID": hospital["HospitalID"],
            "HospitalName": hospital["HospitalName"],
            "ProductID": product["ProductID"],
            "ProductName": product["ProductName"],
            "Category": product["Category"],
            "WarehouseZone": zone,
            "StorageCondition": storage,
            "BatchNumber": f"B{np.random.randint(100000, 999999)}",
            "ManufacturingDate": manufacturing,
            "ExpiryDate": expiry,
            "CurrentStock": current_stock,
            "MinStock": min_stock,
            "MaxStock": max_stock
        })

        inventory_id += 1

# حفظ الملف
inventory_df = pd.DataFrame(inventory)
inventory_df.to_csv("Inventory.csv", index=False)
```

```python
import random
import numpy as np
import pandas as pd

# 1. قراءة البيانات
calendar = pd.read_csv("Calendar.csv")
hospitals = pd.read_csv("Hospitals.csv")
products = pd.read_csv("Products.csv")
suppliers = pd.read_csv("Suppliers.csv")
inventory_df = pd.read_csv("Inventory.csv")

# 2. تجهيز البيانات
inventory_dict = inventory_df.set_index(["HospitalID", "ProductID"]).to_dict("index")
products_dict = products.set_index("ProductID").to_dict("index")

local_suppliers = suppliers[suppliers["SupplierType"] == "Local"].to_dict("records")
imported_suppliers = suppliers[suppliers["SupplierType"] == "Imported"].to_dict("records")

# 3. تهيئة التتبع
open_orders = {}
consumption = []
transactions = []
hospital_requisitions = []  # 1. طلب الاحتياج الموحد للمستشفى (المرحلة الأولى)
purchase_orders = []        # 2. أوامر التوريد الموزعة على الموردين (المرحلة الثانية)
po_line_items = []          # 3. تفاصيل الأصناف

consumption_id = 1
transaction_id = 1
req_id = 1
order_id = 1
line_id = 1

quarter_start_months = [1, 4, 7, 10]

# 4. الـ Loop الرئيسية
for _, day in calendar.iterrows():
    current_date = pd.to_datetime(day["Date"])
    is_quarter_start = (current_date.day == 1 and current_date.month in quarter_start_months)

    # أ. استلام الطلبيات الواصلة
    completed_keys = []
    for key, order_info in open_orders.items():
        if current_date >= order_info["ExpectedDeliveryDate"]:
            inventory_dict[key]["CurrentStock"] += order_info["OrderedQty"]

            transactions.append({
                "TransactionID": f"TRN{transaction_id:06}",
                "Date": current_date,
                "HospitalID": key[0],
                "ProductID": key[1],
                "TransactionType": "Receipt",
                "Quantity": order_info["OrderedQty"],
                "BalanceAfterTransaction": inventory_dict[key]["CurrentStock"]
            })
            transaction_id += 1

            for line in po_line_items:
                if line["OrderID"] == order_info["OrderID"] and line["ProductID"] == key[1]:
                    line["Status"] = "Delivered"

            completed_keys.append(key)

    for key in completed_keys:
        del open_orders[key]

    # ب. الاستهلاك اليومي
    for key, item in inventory_dict.items():
        hospital_id = key[0]
        product_id = key[1]
        product = products_dict[product_id]
        demand = product["DemandLevel"]

        current_stock = item["CurrentStock"]

        if demand == "High":
            daily_qty = random.randint(15, 35)
        elif demand == "Medium":
            daily_qty = random.randint(5, 15)
        else:
            daily_qty = random.randint(1, 6)

        daily_qty = min(daily_qty, current_stock)
        inventory_dict[key]["CurrentStock"] -= daily_qty

        if daily_qty > 0:
            consumption.append({
                "ConsumptionID": f"CON{consumption_id:06}",
                "Date": current_date,
                "HospitalID": hospital_id,
                "ProductID": product_id,
                "QuantityConsumed": daily_qty
            })
            consumption_id += 1

            transactions.append({
                "TransactionID": f"TRN{transaction_id:06}",
                "Date": current_date,
                "HospitalID": hospital_id,
                "ProductID": product_id,
                "TransactionType": "Consumption",
                "Quantity": -daily_qty,
                "BalanceAfterTransaction": inventory_dict[key]["CurrentStock"]
            })
            transaction_id += 1

    # ج. بداية الربع سنة: المستشفى تعمل طلب موحد -> السيستم يوزعه على الموردين
    if is_quarter_start:
        for hosp in hospitals["HospitalID"].unique():
            
            # المرحلة الأولى: المستشفى بتعمل طلب احتياج واحد مجمع لجميع أدوية النقص
            req_code = f"REQ{req_id:05}"
            req_id += 1
            
            items_for_req = []

            for (h_id, p_id), item in inventory_dict.items():
                if h_id == hosp:
                    product = products_dict[p_id]
                    updated_stock = item["CurrentStock"]
                    max_stock = item["MaxStock"]
                    ordered_qty = max_stock - updated_stock

                    if (h_id, p_id) not in open_orders and ordered_qty > 0:
                        items_for_req.append({
                            "ProductID": p_id,
                            "OrderedQty": ordered_qty,
                            "UnitPrice": product["UnitPrice"],
                            "PreferredSupplierType": product["PreferredSupplierType"]
                        })

            if items_for_req:
                # تسجيل طلب المستشفى الموحد الرئيسي
                hospital_requisitions.append({
                    "RequisitionID": req_code,
                    "HospitalID": hosp,
                    "RequisitionDate": current_date,
                    "TotalRequestedItems": len(items_for_req),
                    "Status": "Approved & Allocated"
                })

                # المرحلة الثانية: التوزيعة على الموردين (Award Allocation)
                supplier_groups = {}
                for item in items_for_req:
                    pref_type = item["PreferredSupplierType"]
                    eligible_suppliers = local_suppliers if pref_type == "Local" else imported_suppliers
                    supplier = random.choice(eligible_suppliers)
                    supp_id = supplier["SupplierID"]
                    lead_time = int(supplier["LeadTimeDays"])

                    if supp_id not in supplier_groups:
                        supplier_groups[supp_id] = {
                            "LeadTime": lead_time,
                            "Items": []
                        }
                    
                    supplier_groups[supp_id]["Items"].append(item)

                # إصدار أوامر التوريد لكل مورد بناءً على التوزيعة
                for supp_id, supp_data in supplier_groups.items():
                    po_code = f"PO{order_id:05}"
                    order_id += 1
                    expected_delivery = current_date + pd.Timedelta(days=supp_data["LeadTime"])
                    
                    total_po_cost = sum(i["OrderedQty"] * i["UnitPrice"] for i in supp_data["Items"])

                    purchase_orders.append({
                        "OrderID": po_code,
                        "RequisitionID": req_code, # ربط أمر التوريد بطلب المستشفى الأصلي
                        "HospitalID": hosp,
                        "SupplierID": supp_id,
                        "OrderDate": current_date,
                        "TotalCost": round(total_po_cost, 2),
                        "Status": "Open",
                        "ExpectedDeliveryDate": expected_delivery
                    })

                    for itm in supp_data["Items"]:
                        po_line_items.append({
                            "LineID": f"LINE{line_id:06}",
                            "OrderID": po_code,
                            "ProductID": itm["ProductID"],
                            "OrderedQty": itm["OrderedQty"],
                            "UnitPrice": itm["UnitPrice"],
                            "LineTotal": round(itm["OrderedQty"] * itm["UnitPrice"], 2),
                            "Status": "Open"
                        })
                        line_id += 1

                        open_orders[(hosp, itm["ProductID"])] = {
                            "OrderID": po_code,
                            "OrderedQty": itm["OrderedQty"],
                            "ExpectedDeliveryDate": expected_delivery
                        }

# 5. تحويل البيانات إلى DataFrames
hospital_requisitions_df = pd.DataFrame(hospital_requisitions)
purchase_orders_df = pd.DataFrame(purchase_orders)
po_line_items_df = pd.DataFrame(po_line_items)

print("=" * 60)
print(" تم تحديث المحاكاة لتطابق طلب المستشفى وتوزيع الموردين!")
print("=" * 60)
print(f"1. طلبات المستشفى الموحدة (Requisitions)  : {hospital_requisitions_df.shape[0]} طلبات فقط (طلب لكل مستشفى كل 3 شهور)")
print(f"2. أوامر التوريد للموردين (Purchase Orders): {purchase_orders_df.shape[0]} أمر توريد موزّع")
print(f"3. أسطر وتفاصيل الأصناف (Line Items)        : {po_line_items_df.shape[0]} صنف مطلوبة")
```

```python
import random
import numpy as np
import pandas as pd

# 1. قراءة البيانات
calendar = pd.read_csv("Calendar.csv")
hospitals = pd.read_csv("Hospitals.csv")
products = pd.read_csv("Products.csv")
suppliers = pd.read_csv("Suppliers.csv")
inventory_df = pd.read_csv("Inventory.csv")

# 2. تسريع الأداء بتحويل البيانات لـ Dictionaries
inventory_dict = inventory_df.set_index(["HospitalID", "ProductID"]).to_dict(
    "index"
)
products_dict = products.set_index("ProductID").to_dict("index")

local_suppliers = suppliers[suppliers["SupplierType"] == "Local"].to_dict(
    "records"
)
imported_suppliers = suppliers[suppliers["SupplierType"] == "Imported"].to_dict(
    "records"
)

# 3. تهيئة التتبع والقوائم
open_orders = {}
consumption = []
transactions = []
hospital_requisitions = []  # طلبات المستشفى الموحدة
purchase_orders = []  # أوامر التوريد الرسمية للموردين
po_line_items = []  # أسطر الأصناف والتفاصيل

consumption_id = 1
transaction_id = 1
req_id = 1
order_id = 1
line_id = 1

# أشهُر بداية الأرباع (كل 3 شهور: يناير، أبريل، يوليو، أكتوبر)
quarter_start_months = [1, 4, 7, 10]

# 4. الـ Loop الرئيسية السريعة
for _, day in calendar.iterrows():
    current_date = pd.to_datetime(day["Date"])
    is_quarter_start = (
        current_date.day == 1 and current_date.month in quarter_start_months
    )

    # أ. استلام الطلبيات الواصلة
    completed_keys = []
    for key, order_info in open_orders.items():
        if current_date >= order_info["ExpectedDeliveryDate"]:
            inventory_dict[key]["CurrentStock"] += order_info["OrderedQty"]

            transactions.append(
                {
                    "TransactionID": f"TRN{transaction_id:06}",
                    "Date": current_date,
                    "HospitalID": key[0],
                    "ProductID": key[1],
                    "TransactionType": "Receipt",
                    "Quantity": order_info["OrderedQty"],
                    "BalanceAfterTransaction": inventory_dict[key][
                        "CurrentStock"
                    ],
                }
            )
            transaction_id += 1

            for line in po_line_items:
                if (
                    line["OrderID"] == order_info["OrderID"]
                    and line["ProductID"] == key[1]
                ):
                    line["Status"] = "Delivered"

            completed_keys.append(key)

    for key in completed_keys:
        del open_orders[key]

    # ب. الاستهلاك اليومي
    for key, item in inventory_dict.items():
        hospital_id = key[0]
        product_id = key[1]
        product = products_dict[product_id]
        demand = product["DemandLevel"]

        current_stock = item["CurrentStock"]

        if demand == "High":
            daily_qty = random.randint(15, 35)
        elif demand == "Medium":
            daily_qty = random.randint(5, 15)
        else:
            daily_qty = random.randint(1, 6)

        daily_qty = min(daily_qty, current_stock)
        inventory_dict[key]["CurrentStock"] -= daily_qty

        if daily_qty > 0:
            consumption.append(
                {
                    "ConsumptionID": f"CON{consumption_id:06}",
                    "Date": current_date,
                    "HospitalID": hospital_id,
                    "ProductID": product_id,
                    "QuantityConsumed": daily_qty,
                }
            )
            consumption_id += 1

            transactions.append(
                {
                    "TransactionID": f"TRN{transaction_id:06}",
                    "Date": current_date,
                    "HospitalID": hospital_id,
                    "ProductID": product_id,
                    "TransactionType": "Consumption",
                    "Quantity": -daily_qty,
                    "BalanceAfterTransaction": inventory_dict[key][
                        "CurrentStock"
                    ],
                }
            )
            transaction_id += 1

    # ج. طلبات الشراء المجمعة (في بداية كل ربع سنة)
    if is_quarter_start:
        for hosp in inventory_df["HospitalID"].unique():
            req_code = f"REQ{req_id:05}"
            req_id += 1
            items_for_req = []

            for (h_id, p_id), item in inventory_dict.items():
                if h_id == hosp:
                    product = products_dict[p_id]
                    updated_stock = item["CurrentStock"]
                    max_stock = item["MaxStock"]
                    ordered_qty = max_stock - updated_stock

                    if (h_id, p_id) not in open_orders and ordered_qty > 0:
                        items_for_req.append(
                            {
                                "ProductID": p_id,
                                "OrderedQty": ordered_qty,
                                "UnitPrice": product["UnitPrice"],
                                "PreferredSupplierType": product[
                                    "PreferredSupplierType"
                                ],
                            }
                        )

            if items_for_req:
                hospital_requisitions.append(
                    {
                        "RequisitionID": req_code,
                        "HospitalID": hosp,
                        "RequisitionDate": current_date,
                        "TotalRequestedItems": len(items_for_req),
                        "Status": "Approved & Allocated",
                    }
                )

                supplier_groups = {}
                for item in items_for_req:
                    pref_type = item["PreferredSupplierType"]
                    eligible_suppliers = (
                        local_suppliers
                        if pref_type == "Local"
                        else imported_suppliers
                    )
                    supplier = random.choice(eligible_suppliers)
                    supp_id = supplier["SupplierID"]
                    lead_time = int(supplier["LeadTimeDays"])

                    if supp_id not in supplier_groups:
                        supplier_groups[supp_id] = {
                            "LeadTime": lead_time,
                            "Items": [],
                        }

                    supplier_groups[supp_id]["Items"].append(item)

                for supp_id, supp_data in supplier_groups.items():
                    po_code = f"PO{order_id:05}"
                    order_id += 1
                    expected_delivery = current_date + pd.Timedelta(
                        days=supp_data["LeadTime"]
                    )
                    total_po_cost = sum(
                        i["OrderedQty"] * i["UnitPrice"]
                        for i in supp_data["Items"]
                    )

                    purchase_orders.append(
                        {
                            "OrderID": po_code,
                            "RequisitionID": req_code,
                            "HospitalID": hosp,
                            "SupplierID": supp_id,
                            "OrderDate": current_date,
                            "TotalCost": round(total_po_cost, 2),
                            "Status": "Open",
                            "ExpectedDeliveryDate": expected_delivery,
                        }
                    )

                    for itm in supp_data["Items"]:
                        po_line_items.append(
                            {
                                "LineID": f"LINE{line_id:06}",
                                "OrderID": po_code,
                                "ProductID": itm["ProductID"],
                                "OrderedQty": itm["OrderedQty"],
                                "UnitPrice": itm["UnitPrice"],
                                "LineTotal": round(
                                    itm["OrderedQty"] * itm["UnitPrice"], 2
                                ),
                                "Status": "Open",
                            }
                        )
                        line_id += 1

                        open_orders[(hosp, itm["ProductID"])] = {
                            "OrderID": po_code,
                            "OrderedQty": itm["OrderedQty"],
                            "ExpectedDeliveryDate": expected_delivery,
                        }

# 5. تحويل البيانات إلى DataFrames
df_consumption = pd.DataFrame(consumption)
df_transactions = pd.DataFrame(transactions)
df_requisitions = pd.DataFrame(hospital_requisitions)
df_purchase_orders = pd.DataFrame(purchase_orders)
df_po_lines = pd.DataFrame(po_line_items)

# 6. حفظ البيانات
df_consumption.to_csv("Consumption.csv", index=False)
df_transactions.to_csv("Inventory_Transactions.csv", index=False)
df_purchase_orders.to_csv("Purchase_Orders.csv", index=False)
df_po_lines.to_csv("Purchase_Order_Lines.csv", index=False)

# 7. طباعة المخرجات والنتائج فوراً على الشاشة!
print("=" * 60)
print(" تم تشغيل المحاكاة واستخراج البيانات بنجاح!")
print("=" * 60)
print(f"📊 عدد صفوف الاستهلاك (Consumption)  : {df_consumption.shape[0]}")
print(f"📊 عدد صفوف الحركات (Transactions)   : {df_transactions.shape[0]}")
print(f"📊 طلبات المستشفيات (Requisitions)   : {df_requisitions.shape[0]}")
print(f"📊 أوامر التوريد للموردين (POs)      : {df_purchase_orders.shape[0]}")
print(f"📊 أسطر الأصناف المجمعة (PO Lines)    : {df_po_lines.shape[0]}")
print("=" * 60)

# عرض أول 5 صفوف من جدول أوامر الشراء
display(df_purchase_orders.head())
```
